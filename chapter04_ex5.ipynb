```python
from jupyturtle import forward, left, make_turtle, right
import math
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)

def arc(radius, angle, n = 30):
    arc_length = 2 * math.pi * radius * angle /360

    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)

def petal(radius, angle):
    for i in range(2):
        arc(radius, angle)
        left(180 - angle)

def draw_flower(radius, petals):
    angle = 360 / petals
    for i in range(petals):
        petal(radius, angle)
        right(angle)

make_turtle(delay = 0.0001)
draw_flower(20, 5)
```
