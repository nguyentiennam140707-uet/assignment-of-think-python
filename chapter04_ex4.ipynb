```python
from jupyturtle import left, right, forward, make_turtle
import math
def triangle(length, angle):
    base_angle = (180 - angle) / 2
    base_length = math.sin(math.radians(angle/2)) * length * 2
    forward(length)
    left(180 - base_angle)
    forward(base_length)
    left(180 - base_angle)
    forward(length)


def draw_pie(length, segments):
    for i in range(segments):
        angle = 360 / segments
        triangle(length, angle)
        left(180)

make_turtle()
draw_pie(50, 10)
```
